```python
from dotenv import load_dotenv
import os

from alpaca.data.historical import StockHistoricalDataClient
from alpaca.data.requests import StockBarsRequest
from alpaca.data.timeframe import TimeFrame
from alpaca.trading.client import TradingClient
```

```python

load_dotenv()

KEY = os.getenv("key")
SECRET = os.getenv("secret")
```

```python
client = StockHistoricalDataClient(KEY, SECRET)

request = StockBarsRequest(
    symbol_or_symbols = "AAPL",
    timeframe = TimeFrame.Day,
    start = "2025-01-01",
)
bars = client.get_stock_bars(request)

df = bars.df

df
```

```python
trader = TradingClient(KEY, SECRET)

bruker = trader.get_account()

print(bruker.equity) #equity = egenkapittal
print(bruker.cash)  #mien penger
print(bruker.portfolio_value) #total verdi av porteføljen
print(bruker.buying_power) #total verdi av porteføljen
```

```python
positions = trader.get_all_positions()

for position in positions:
    print(f"Symbol: {position.symbol}, Quantity: {position.qty}, Market Value: {position.market_value}")
    #får ingenting ut fordi jeg har ikke kjøpt noen aksjer på allpacka
```
