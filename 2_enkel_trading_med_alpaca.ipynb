```python
import os
from dotenv import load_dotenv

from alpaca.data.historical import StockHistoricalDataClient
from alpaca.data.requests import StockBarsRequest
from alpaca.trading.requests import OrderRequest
from alpaca.trading.client import TradingClient
from alpaca.data.timeframe import TimeFrame
```

```python
load_dotenv()

KEY = os.getenv("key")
SECRET = os.getenv("secret")
```

```python

trader = TradingClient(KEY, SECRET)
request = OrderRequest(
    symbol="AAPL",
    qty=1,
    side="buy",
    type="market",
    time_in_force="gtc"
)

trader.submit_order(request)

print(trader.get_orders())
```

```python
client = StockHistoricalDataClient(KEY, SECRET)

def get_close(symbol):
    request = StockBarsRequest(
        symbol_or_symbols=symbol,
        timeframe=TimeFrame.Day,
        start="2026-01-01",
    )
    bars = client.get_stock_bars(request)
    df = bars.df
    return df["close"]

def strategy(symbol):
    close_prices = get_close(symbol)
    last_close = close_prices.iloc[-1]
    second_last_close = close_prices.iloc[-2]

    direction = last_close - second_last_close
    threshold = direction * 0.05

    if direction >= threshold:
        return "buy"
    elif direction <= threshold:
        return "sell"
    else:
        return "hold"
```

```python
print(strategy("AAPL"))
```
